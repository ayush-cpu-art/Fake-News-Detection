```python
print("Statement duplicates:", df["Statement"].duplicated().sum())
```

```python
duplicate_statements = df[df["Statement"].duplicated(keep=False)]

label_conflicts = (duplicate_statements.groupby("Statement")["Label"].nunique())

print("Duplicate statements:", duplicate_statements["Statement"].nunique())
print("Statements with conflicting labels:", (label_conflicts > 1).sum())
```

```python
conflicting_statements = label_conflicts[label_conflicts > 1].index

conflicting_rows = df[df["Statement"].isin(conflicting_statements)]

print("Conflicting statements:", len(conflicting_statements))
print("Rows affected:", len(conflicting_rows))
```

```python
display(conflicting_rows[["Statement", "Label"]].sort_values("Statement").head(20))
```

```python
conflicting_statements = label_conflicts[label_conflicts > 1].index

clean_df = df[~df["Statement"].isin(conflicting_statements)].copy()

print("Rows after removing conflicting statements:", len(clean_df))
print("Remaining statement duplicates:", clean_df["Statement"].duplicated().sum())
```

```python
clean_df = clean_df.drop_duplicates(subset="Statement", keep="first").copy()

print("Final rows after duplicate removal:", len(clean_df))
```

```python
print("Final label distribution:")
print(clean_df["Label"].value_counts())
```
